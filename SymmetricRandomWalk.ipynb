### Performing a symmetric random walk on a dataset of graphs using NetworkX

The function symmetric_random_walk performs a symmetric random walk on the graphs in dataset ENZYME. It starts the random walk from a random node and selects the next node based on equal probabilities for all neighboring nodes. The walk continues for a specified custom number of steps (num_steps). 

```python
import numpy as np
import networkx as nx

def symmetric_random_walk(graph, num_steps):
    
    ##Perform a symmetric random walk on the given graph.
    # Start the random walk from a random node
    current_node = np.random.choice(list(graph.nodes()))

    walk = [current_node]

    for _ in range(num_steps - 1):
        neighbors = list(graph.neighbors(current_node))
        if neighbors:
            next_node = np.random.choice(neighbors)
            walk.append(next_node)
            current_node = next_node
        else:
            # If no neighbors are available, terminate the walk
            break

    return walk

# Example usage:
# Create a NetworkX graph


# Perform a symmetric random walk on the graph
random_walk = symmetric_random_walk(G1, num_steps=10)
print("Random walk:", random_walk)
```
