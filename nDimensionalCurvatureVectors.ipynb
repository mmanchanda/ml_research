```python
import numpy as np
import networkx as nx
from grakel import GraphKernel
import matplotlib.pyplot as plt
import seaborn as sns
from torch_geometric.utils import to_networkx
from sklearn.metrics.pairwise import cosine_similarity
```

```python
from torch_geometric.datasets import TUDataset
tu_data = TUDataset(root="tutorial1",name= "Yeast")
# https://pytorch-geometrimport networkx as nx
t1 = tu_data[1]
len(t1)
```

```python
G1 = to_networkx(t1, node_attrs=["x"], edge_attrs=["edge_attr"])
```

```python
nx.draw(G1, with_labels = True)
plt.show()
```

### Compute Curvature Vectors

```python
curvature_vectors = []
for node in G1.nodes():
    degree = G1.degree[node]
    curvature_vector = [degree]  # You can extend this with more metrics
    curvature_vectors.append(curvature_vector)
```

### Compute Cosine Similarity

```python
similarities = cosine_similarity(curvature_vectors).reshape(-1, 1)

average_similarity = np.mean(similarities)

print("Average Cosine Similarity:", average_similarity)
```
